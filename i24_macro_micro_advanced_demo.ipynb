```python
import simulation
import json
```

```python
with open("i24_motion_to_dataset.json", "r") as f:
    config = json.load(f)
```

```python
sim = simulation.Simulation.from_json(
    json_path="simulation_dataset/network.json",
    time_resolution=1.0,
    origin_time=config["time_origin"],
    free_flow_speed=30.0,
    congestion_wave_speed=5.0,
    jam_density=0.16,
)
gt = simulation.GroundTruthStore.from_parquet("simulation_dataset/micro.parquet", "simulation_dataset/macro.parquet")
sim.initialize_from_ground_truth(gt, time_value=config["time_origin"])
```
