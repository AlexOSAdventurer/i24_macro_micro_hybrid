```python
import i24_motion_macro
import matplotlib.pyplot as plt
import numpy
```

```python
i24macro = i24_motion_macro.I24MotionMacro(i24_motion_macro.i24_motion_data.I24MotionData(2, 1669812350, 1669812350+3600, 0, 1600), 2, "road_2")
```

```python
raw_macro = i24macro.loadRawMacroData()
processed_macro = i24macro.loadProcessedMacroData()
queries = i24macro.computeEdieBoxQueries()
```

```python
processed_macro_lane_1 = processed_macro[-1]
raw_macro_lane_1 = raw_macro[-1]
```

```python
timestamp_min_display = numpy.array(queries["timestamp_min"]).reshape((7200, 80)).T
s_min_display = numpy.array(queries["s_min"]).reshape((7200, 80)).T
density_processed = (processed_macro_lane_1["density"]).reshape(7200, 80).T
density_raw = (raw_macro_lane_1["density"]).reshape(7200, 80).T
print(density_processed.shape)
plt.pcolormesh(timestamp_min_display, s_min_display, density_processed, cmap='viridis', shading='auto')
plt.title('Density - 1 Hour, 1 Mile')
plt.colorbar(label='Density')
plt.show()
plt.pcolormesh(timestamp_min_display, s_min_display, density_raw, cmap='viridis', shading='auto')
plt.title('Density - 1 Hour, 1 Mile')
plt.colorbar(label='Density')
plt.show()
```

```python
timestamp_min_display = numpy.array(queries["timestamp_min"]).reshape((7200, 80)).T
s_min_display = numpy.array(queries["s_min"]).reshape((7200, 80)).T
velocity_processed = (processed_macro_lane_1["velocity"]).reshape(7200, 80).T
velocity_raw = (raw_macro_lane_1["velocity"]).reshape(7200, 80).T
print(velocity_processed.shape)
plt.pcolormesh(timestamp_min_display, s_min_display, velocity_processed, cmap='viridis', shading='auto')
plt.title('Velocity - 1 Hour, 1 Mile')
plt.colorbar(label='Velocity')
plt.show()
plt.pcolormesh(timestamp_min_display, s_min_display, velocity_raw, cmap='viridis', shading='auto')
plt.title('Velocity - 1 Hour, 1 Mile')
plt.colorbar(label='Velocity')
plt.show()
```

```python
timestamp_min_display = numpy.array(queries["timestamp_min"]).reshape((7200, 80)).T
s_min_display = numpy.array(queries["s_min"]).reshape((7200, 80)).T
flow_processed = (processed_macro_lane_1["flow"]).reshape(7200, 80).T
flow_raw = (raw_macro_lane_1["flow"]).reshape(7200, 80).T
print(velocity_processed.shape)
plt.pcolormesh(timestamp_min_display, s_min_display, flow_processed, cmap='viridis', shading='auto')
plt.title('Flow - 1 Hour, 1 Mile')
plt.colorbar(label='Flow')
plt.show()
plt.pcolormesh(timestamp_min_display, s_min_display, flow_raw, cmap='viridis', shading='auto')
plt.title('Flow - 1 Hour, 1 Mile')
plt.colorbar(label='Flow')
plt.show()
```
