```python
import simulation
import json
import os
```

```python
with open("i24_motion_to_dataset.json", "r") as f:
    config = json.load(f)

sim = simulation.Simulation.from_json(
     json_path=os.path.join(config["storage_locations"]["simulation_dataset"], "network.json"),
     time_resolution=config["time_step"],
     origin_time=config["time_origin"]
)

gt = simulation.GroundTruthStore.from_parquet(os.path.join(config["storage_locations"]["simulation_dataset"], "micro.parquet"), os.path.join(config["storage_locations"]["simulation_dataset"], "macro.parquet"))
sim.initialize_from_ground_truth(gt, time_value=config["time_origin"])
```

```python
sim.run(duration=360.0)
```

```python
sim.plot_rollout(rotation_deg=90.0)
```
